Створіть нову таблицю python_data, в якій будуть тільки рядки зі спеціалістами, які вказали мову програмування Python

```python
python_data = df[df['Мова програмування'] == 'Python']
```

Визначте розмір таблиці python_data за допомогою методу shape

```python
python_data.shape
```

Використовуючи метод groupby, виконайте групування за стовпчиком "Посада"

```python
grouped_by_position = python_data.groupby('Посада')
grouped_by_position.size()
```

Створіть новий DataFrame, де для згрупованих даних за стовпчиком "Посада", виконайте агрегацію даних за допомогою методу agg і знайдіть мінімальне та максимальне значення у стовпчику "Зарплата.в.місяць"

```python
salary_agg = python_data.groupby('Посада')['Зарплата на місяць'].agg(['min', 'max'])
salary_agg
```

Створіть функцію fill_avg_salary, яка повертатиме середнє значення заробітної плати на місяць. Використовуйте її для методу apply та створіть новий стовпчик "avg"

```python
def fill_avg_salary(row):
    return (row['min'] + row['max']) / 2

salary_agg['avg'] = salary_agg.apply(fill_avg_salary, axis=1)
salary_agg
```

Створіть описову статистику за допомогою методу describe для нового стовпчика.

```python
avg_stats = salary_agg['avg'].describe()
avg_stats
```

Додайте від 3 до 5 графіків функцій різного типу на ваш вибір. Задайте графікам оформлення, щоб кожен графік у своїй домашній роботі чимось відрізнявся і не був схожим на інші. Можна використовувати як matplotlib, так і seaborn.

```python
x = np.linspace(-10, 10, 200)

plt.figure(figsize=(8, 4))
plt.plot(x, x**2, color="magenta", lw=2)
plt.scatter([0], [0], color="orange", s=100, zorder=5, label='Вершина')
plt.title("Квадратична функція", weight='bold')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(x, x**3, color="black", ls=':', lw=2)
plt.grid(color='gray', alpha=0.2)
plt.title("Кубічна парабола")
plt.show()
```
